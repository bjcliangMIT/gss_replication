```python
from os import listdir
import pandas as pd
import random
import seaborn as sns
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
def isfloat(num):
    try:
        float(num); return True
    except ValueError: return False
```

```python
def get_significant_vars(data, index):
    row = data.iloc[index]
    #assert(row['type'].lower()=='original')
    pvalues = [row[i] for i in dict(row) if 'pvalue' in i]
    #print(pvalues)
    k = [i for i in pvalues if (type(i)==str) and (not isfloat(i))]
    #print(k)
    if len(k)>0:
        significant = []
        for i in pvalues:
            if i != i: significant.append(False)
            elif '*' in i: significant.append(True)
            else: print(i)
        significant = np.array(significant)
    else:
        pvalues = [float(i) for i in pvalues]
        significant = np.array(pvalues)<=0.05
    #print(significant)
    significant = sum(significant)/len(pvalues)
    
    if 'dependent' in row: 
        return row['dependent'], row['type'], significant
    return np.nan, row['type'], significant
```
